```python
import numpy as np

y_predicted = np.array([1, 1, 0, 0, 1])
y_true = np.array([0.30, 0.7, 1, 0, 0.5])
```

```python
def mea(y_true, y_predicted):
    total_error = 0
    for yt, yp in zip(y_true, y_predicted):
        total_error += abs(yt - yp)
    print("Total error: ", total_error)
    mea = total_error / len(y_true)
    print("Mean absolute error: ", mea)
    return mea
```

```python
mea(y_true, y_predicted)
```

```python
np.mean(np.abs(y_predicted - y_true))
```

```python
def log_loss(y_true, y_predicted):
    epsilon = 1e-15
    y_predicted_new = np.array([max(i, epsilon) for i in y_predicted])
    y_predicted_new = np.array([min(i, 1 - epsilon) for i in y_predicted_new])
    return -np.mean(y_true * np.log(y_predicted_new) + (1 - y_true) * np.log(1 - y_predicted_new))
```
