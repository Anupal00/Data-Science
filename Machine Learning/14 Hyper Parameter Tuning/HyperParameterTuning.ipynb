```python
import numpy as np
from sklearn import svm, datasets
iris = datasets.load_iris()
```

```python
import pandas as pd
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

```python
df["flower"] = iris.target
df["flower"] = df["flower"].apply(lambda x: iris.target_names[x])
df[47:52]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size = 0.2)
```

```python
model = svm.SVC(kernel='rbf', C=30, gamma="auto")
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(svm.SVC(kernel="linear", C=10, gamma="auto"), iris.data, iris.target, cv=5)
```

```python
cross_val_score(svm.SVC(kernel="rbf", C=10, gamma="auto"), iris.data, iris.target, cv=5)
```

```python
cross_val_score(svm.SVC(kernel="rbf", C=20, gamma="auto"), iris.data, iris.target, cv=5)
```

```python
kernels = ["rbf", "linear"]
C = [1, 10, 20]
avg_score = {}
for kernel in kernels:
    for c in C:
        cv_scores = cross_val_score(svm.SVC(kernel=kernel, C=c, gamma = "auto"), iris.data, iris.target, cv=5)
        avg_score[kernel + "_" + str(c)] = np.average(cv_scores)
avg_score
```

```python
from sklearn.model_selection import GridSearchCV
clf = GridSearchCV(svm.SVC(gamma = "auto"), {
    "C": [1, 10, 20],
    "kernel": ["rbf", "linear"]
}, cv=5, return_train_score=False)
clf.fit(iris.data, iris.target)
clf.cv_results_
```

```python
df = pd.DataFrame(clf.cv_results_)
df
```

```python
df[["param_C", "param_kernel", "mean_test_score"]]
```

```python
clf.best_score_
```

```python
clf.best_params_
```

```python
from sklearn.model_selection import RandomizedSearchCV
rs = RandomizedSearchCV(svm.SVC(gamma="auto"), {
    "C": [1, 10, 20],
    "kernel": ["rbf", "linear"]
}, cv=5, n_iter=2, return_train_score=False)
rs.fit(iris.data, iris.target)
pd.DataFrame(rs.cv_results_)
```

```python
from sklearn import svm
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

model_params = {
    "svm": {
        "model": svm.SVC(gamma="auto"),
        "params": {
            "C": [1, 10, 20],
            "kernel": ["rbf", "linear"]
        }
    },
    "random_forest": {
        "model": RandomForestClassifier(),
        "params": {
            "n_estimators": [1, 5, 10],
        }
    },
    "logistic_regression": {
        "model": LogisticRegression(solver="liblinear", multi_class="auto", max_iter=100),
        "params": {
            "C": [1, 5, 10],
        }
    }
}
```

```python
scores = []
for model_name, mp in model_params.items():
    clf = GridSearchCV(mp["model"], mp["params"], cv=5, return_train_score=False)
    clf.fit(iris.data, iris.target)
    scores.append({
        "model_name": model_name,
        "best_score": clf.best_score_,
        "best_params": clf.best_params_,
    })
scores
```

```python
df = pd.DataFrame(scores)
```
