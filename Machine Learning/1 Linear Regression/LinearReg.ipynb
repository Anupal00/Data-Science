```python
import matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv('homeprices.csv')
print(df.head())
```

```python
%matplotlib inline
plt.title("Scatterplot of area vs price")
plt.xlabel("area")
plt.ylabel("price")
plt.scatter(df['area'], df['price'], color='red', marker='o')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[["area"]], df.price)
```

```python
reg.predict([[3300]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
d = pd.read_csv("areas.csv")
```

```python
r = reg.predict(d)
```

```python
d["price"] = r
```

```python
print(d.columns)
```

```python
d.to_csv("prediction.csv", index = False)
```

```python
%matplotlib inline
plt.title("Scatterplot of area vs price")
plt.xlabel("area")
plt.ylabel("price")
plt.scatter(df['area'], df['price'], color='red', marker='o')
plt.plot(df['area'], reg.predict(df[["area"]]), color="blue")
```
